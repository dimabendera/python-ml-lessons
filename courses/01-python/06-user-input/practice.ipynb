## 0 · Чому в зошиті немає справжнього `input()`

Зошит виконується не в терміналі. Між твоїм кодом і клавіатурою стоїть ядро Jupyter,
і при автоматичному прогоні (саме так перевіряють цей файл) відповідати на запит нікому.

`input()` у такому режимі **не зависає** — він падає. Ось як саме. Ця клітинка навмисно
з помилкою: подивись на назву винятку й на текст повідомлення.

````python
# навмисна помилка: показуємо, що саме станеться з input() у зошиті
name = input("Як тебе звати? ")
print(name)
````

Тому далі ми **не викликаємо `input()`**, а записуємо відповідь самі — рівно ту,
яку він повернув би. Це не милиця, а сама суть теми: `input()` привʼязує програму
до людини за клавіатурою, і саме тому його не можна ані автоматизувати, ані перевірити.

У розділі 8 ми все ж запустимо справжній `input()` — але вже у власному файлі `.py`,
який зошит запустить окремим процесом і нагодує заготовленими відповідями.

````python
# рівно те, що повернув би input("Як тебе звати? ")
name = "Олена"
# рівно те, що повернув би input("Скільки тобі років? ")
age_text = "14"
# рівно те, що повернув би input("Скільки зошитів? ")
count_text = "3"

print("name       =", repr(name), "  тип:", type(name))
print("age_text   =", repr(age_text), "     тип:", type(age_text))
print("count_text =", repr(count_text), "      тип:", type(count_text))
````

## 1 · Головна пастка: `input()` завжди повертає рядок

Зверни увагу на попередній вивід: тип усіх трьох змінних — `str`. Навіть у тієї,
де на екрані цифри. Подивимось, що з цього випливає.

````python
# плюс для двох рядків означає склеювання, а не додавання
print('"2" + "2" дає', repr("2" + "2"), "— рядок довжиною", len("2" + "2"))
print(" 2  +  2  дає", 2 + 2, "— число")
print()

# а множення рядка на число — цілком законна операція з теми 05
print("age_text * 2 дає", repr(age_text * 2), "— і жодної помилки при цьому немає")

assert age_text + age_text == "1414", "склеювання рядків працює не так, як ми думали"
assert int(age_text) + int(age_text) == 28
print()
print("✅ те саме «плюс» дає 1414 для рядків і 28 для чисел")
````

### Помилка, яка падає, — це щастя

`"14" * 2` мовчки дало неправильний результат. А от спроба додати число до рядка
зупиняє програму одразу. Клітинка нижче знову навмисно з помилкою.

````python
# навмисна помилка: рядок і число не складаються
print(age_text + 1)
````

## 2 · Перетворення: `int()` і `float()`

Ліки одні: перетворити текст на число явно. Подивимось, що ці функції приймають,
а що ні.

````python
age = int(age_text)
count = int(count_text)

print("int('14')   ->", int("14"))
print("int('  14 ')->", int("  14 "), "  — пробіли по краях дозволені")
print("int('-7')   ->", int("-7"), "   — знак дозволений")
print("int('1_000')->", int("1_000"), " — підкреслення як розділювач теж дозволене")
print("float('42.5') ->", float("42.5"))
print("int(float('42.5')) ->", int(float("42.5")), " — дробова частина ВІДКИДАЄТЬСЯ")
print("round(42.5) ->", round(42.5), "  — а це вже округлення, і воно інше")
print()
print("вік через рік:", age + 1)

assert age == 14 and isinstance(age, int)
assert int(float("42.9")) == 42, "int() від дробового має відкидати, а не округлювати"
print()
print("✅ int() відкидає дробову частину, round() округлює — це різні дії")
````

### А якщо людина ввела «сорок»

Ось справжній traceback. Прочитай його знизу вгору: тип помилки, потім пояснення,
потім те, що саме ми дали функції.

````python
# навмисна помилка: «сорок» не є записом числа
print(int("сорок"))
````

Полагодити це — тобто перепитати замість того, щоб упасти — ми зможемо лише після
трьох тем: умов (12), циклів (13) і винятків (19). Поки що правильна поведінка
програми на такому вводі — впасти гучно.

Подивимось ще на кілька випадків, де межа проходить не там, де здається.

````python
# циклів і списків ми ще не проходили, тому перевіряємо кожен випадок окремим рядком
print("int('42')     працює ->", int("42"))
print("float('42.5') працює ->", float("42.5"))
print()
print("а ось що НЕ працює (текст помилки — з реального запуску):")
print("  int('42.5')    -> ValueError: invalid literal for int() with base 10: '42.5'")
print("  int('42 000')  -> ValueError: invalid literal for int() with base 10: '42 000'")
print("  float('3,5')   -> ValueError: could not convert string to float: '3,5'")
print()
print("правило: int() хоче суцільні цифри, float() дозволяє КРАПКУ й нічого більше")
````

## 3 · Пробіли й регістр

Людина тисне пробіл і пише «Так» з великої. `input()` знімає лише Enter —
усе решта лишається в рядку.

````python
# те, що повернув би input("Продовжити? (так/ні): ") після неохайного вводу
raw_answer = "  Так  "

print("як є:              ", repr(raw_answer), " довжина", len(raw_answer))
print("після .strip():    ", repr(raw_answer.strip()), "     довжина", len(raw_answer.strip()))
print("після .strip().lower():", repr(raw_answer.strip().lower()))
print()
print('порівняння без чистки: raw_answer == "так" ->', raw_answer == "так")
print('порівняння з чисткою:  raw_answer.strip().lower() == "так" ->',
      raw_answer.strip().lower() == "так")

assert raw_answer != "так", "без чистки рядки не збігаються — саме про це й розділ"
assert raw_answer.strip().lower() == "так"
print()
print("✅ обидва рядки виглядають однаково на екрані, а рівні лише після чистки")
````

### `.strip()` працює тільки з краями

Це не недогляд, а свідоме рішення: інакше зникли б пробіли всередині імен із двох слів.

````python
inner_spaces = "т а к"
two_words = "  Марія Петренко  "

print(repr(inner_spaces.strip().lower()), '— пробіл усередині лишився, і це НЕ "так"')
print(repr(two_words.strip()), "— а тут пробіл усередині якраз потрібен")

assert inner_spaces.strip().lower() != "так"
assert two_words.strip() == "Марія Петренко"
print()
print("✅ .strip() чіпає лише краї — і саме тому імена з двох слів не псуються")
````

## 4 · `print` глибше: `sep` і `end`

`print` приймає скільки завгодно значень, склеює їх через `sep` (за замовчуванням пробіл)
і закінчує через `end` (за замовчуванням перехід на новий рядок).

````python
print("Ціна:", 24.5, "грн")            # між значеннями пробіл зʼявився сам
print("а", "б", "в", sep="")           # склеїли впритул
print("а", "б", "в", sep=" · ")        # свій роздільник
print()                                # порожній рядок — найдешевший розділювач блоків

# end="" не переводить рядок, тому наступний print продовжує цей самий
print("рахуємо", end="")
print("...", end="")
print(" готово")

# число print перетворює на текст сам — на відміну від плюса
print("вік:", age)
print()
print('а от "вік: " + age впаде з TypeError — плюс так не вміє')
````

### Де `sep` програє f-рядку

`sep` один на **всі** проміжки одразу. Якщо в рядку потрібні різні проміжки —
допоможе тільки f-рядок.

````python
# три значення й один sep: пробіл зʼявився і там, де він не потрібен
print("через sep:     ", end="")
print("Привіт,", name, "!")
print("через f-рядок:  " + f"Привіт, {name}!")
print()

# рівно те, що зробить print із трьома значеннями й типовим sep
with_sep = "Привіт," + " " + name + " " + "!"
with_fstring = f"Привіт, {name}!"

print("довжина через sep:     ", len(with_sep))
print("довжина через f-рядок: ", len(with_fstring))
assert with_sep != with_fstring, "sep і f-рядок мали б дати різний результат"
assert len(with_sep) - len(with_fstring) == 1
print()
print("✅ різниця рівно в одному символі — зайвому пробілі перед знаком оклику")
````

## 5 · Число у виводі: `.2f` і розділювач тисяч

Тут головне зрозуміти одну річ: **форматування виготовляє текст, а не нове число**.
Перевіримо це руками.

````python
price = 24.50
total = price * count          # 24.5 * 3

print("як надрукує звичайний print:", total)
print("з двома знаками після крапки:", f"{total:.2f}")
print()
print("це РЯДОК:", repr(f"{total:.2f}"), " тип:", type(f"{total:.2f}"))
print("а сама змінна не змінилась:", total, " тип:", type(total))

# round() дає ЧИСЛО і не дописує нуль, .2f дає ТЕКСТ і дописує
assert str(round(total, 2)) == "73.5", "round лишає число, і хвостовий нуль йому не потрібен"
assert f"{total:.2f}" == "73.50", ".2f має дописати нуль до двох знаків"
print()
print("✅ round(73.5, 2) друкується як 73.5, а f'{73.5:.2f}' — як 73.50")
````

### Свій розділювач тисяч проти вбудованого

Скільки секунд прожила чотирнадцятирічна людина? Порахуємо — і розставимо розряди
спершу руками, зрізами з теми 05, а потім міні-мовою формату. Числа мають збігтися.

````python
seconds = age * 365 * 24 * 3600
digits = str(seconds)

print("секунд прожито:", seconds, " цифр у числі:", len(digits))

# рівно три групи по три цифри — тому можна взяти зрізами, без жодного циклу
by_hand = digits[0:3] + "," + digits[3:6] + "," + digits[6:9]
by_python = f"{seconds:,}"

print("наш розбір на розряди:  ", by_hand)
print("міні-мова формату :,   ", by_python)
print("вона ж з підкресленням: ", f"{seconds:_}")
print("а українською пишуть:   ", f"{seconds:,}".replace(",", " "))

assert by_hand == by_python, "наш розбір на розряди розійшовся з вбудованим"
print()
print("✅ усередині :, немає магії — це та сама розстановка роздільників кожні три цифри")
````

## 6 · Вирівнювання: свій відступ проти `format`

Ширина поля — це «допиши пробілів, поки не набереться N символів». Напишемо це
руками й зіставимо з тим, що робить міні-мова.

````python
width = 16
item = "рюкзак шкільний"

# те саме, що робить :<16 — дописати пробілів до потрібної ширини
by_hand_left = item + " " * (width - len(item))
by_python_left = format(item, "<16")

print("наше вирівнювання: ", repr(by_hand_left))
print("format(item, '<16'):", repr(by_python_left))
print()
print("праворуч: ", repr(format(item, ">16")))
print("по центру:", repr(format(item, "^16")))
print("з крапками:", repr(format(item, ".<16")))

assert by_hand_left == by_python_left, "наше вирівнювання розійшлося з format"
print()
print("✅ вирівнювання — це просто дописані символи, і жодної іншої магії там немає")
````

### Ширина — це мінімум, а не максимум

Значення довше за поле нікуди не обріжеться. Саме через це розсипаються таблиці.

````python
narrow = format(item, "<10")           # поле вужче за назву

print("поле 10, назва", len(item), "символів ->", repr(narrow))
print("довжина результату:", len(narrow), "— поле не втримало")

assert len(narrow) == len(item), "Python не має обрізати значення, він розсуває рядок"
assert len(format("зошит", "<10")) == 10, "а коротке значення таки доповнюється до 10"
print()
print("✅ ширина — мінімум: коротке доповнюється, довге лишається як є")
````

## 7 · Складаємо чек

Тепер усе разом: назва ліворуч, сума праворуч, доріжка крапок між ними.

## 8 · Справжній `input()` — у справжньому скрипті

А тепер найцікавіше. Напишемо файл `hello.py` у тимчасову теку й запустимо його
**окремим процесом** — так само, як це зробив би термінал. Там `input()` працює
по-справжньому, а відповіді ми подамо йому наперед, як через конвеєр.

````python
import subprocess
import sys
import tempfile
import shutil
import pathlib

temp_dir = pathlib.Path(tempfile.mkdtemp())     # прибираємо її наприкінці розділу
script_path = temp_dir / "hello.py"

script_text = '''name = input("Як тебе звати? ").strip()
age = int(input("Скільки тобі років? "))
count = int(input("Скільки зошитів купуєш? "))

price = 24.50
total = price * count
seconds = age * 365 * 24 * 3600

print()
print(f"Привіт, {name}!")
print(f"Через рік тобі буде {age + 1}.")
print(f"Ти прожив(ла) вже {seconds:,} секунд.")
print(f"Зошитів: {count} шт. по {price:.2f} грн — разом {total:.2f} грн.")
'''

script_path.write_text(script_text, encoding="utf-8")
print("файл записано:", script_path)
print("рядків у ньому:", len(script_text.splitlines()))
````

Запускаємо. Аргумент `input=` подає процесу три рядки — рівно те саме, що робить
`printf 'Олена\n14\n3\n' | python3 hello.py` у терміналі.

````python
run = subprocess.run(
    [sys.executable, str(script_path)],
    input="Олена\n14\n3\n",          # три відповіді, кожна з новим рядком, як від Enter
    capture_output=True, text=True)

print("код завершення:", run.returncode)
print("--- те, що надрукувала програма ---")
print(run.stdout)

assert run.returncode == 0, "скрипт мав відпрацювати без помилок"
assert "Привіт, Олена!" in run.stdout
assert "441,504,000" in run.stdout
print("✅ справжній input() спрацював — просто джерелом клавіатури став наш рядок")
````

Зверни увагу на перший рядок виводу: підказки «Як тебе звати? », «Скільки тобі років? »
і «Скільки зошитів купуєш? » стоять поспіль одна за одною. Так і має бути:
`input()` друкує підказку **без переходу на новий рядок**, а відповідь у нашому випадку
прийшла не з клавіатури, тому й не відбилась на екрані.

### А що буде, якщо відповідей немає

Порожній вхід — це не порожній рядок. Це `EOFError`.

````python
empty_run = subprocess.run(
    [sys.executable, str(script_path)],
    input="",                        # жодної відповіді
    capture_output=True, text=True)

print("код завершення:", empty_run.returncode, " (нуль означав би успіх)")
print("--- останній рядок помилки ---")
print(empty_run.stderr.strip().splitlines()[-1])

assert empty_run.returncode != 0
assert "EOFError" in empty_run.stderr
print()
print("✅ EOFError означає «джерела вводу більше немає» — це не те саме, що порожня відповідь")
````

### Не вистачило однієї відповіді

Найпідступніший випадок: перші два питання отримали відповідь, а третє — ні.
Програма встигла попрацювати й аж потім упала.

````python
short_run = subprocess.run(
    [sys.executable, str(script_path)],
    input="Олена\n14\n",             # на третє питання відповіді немає
    capture_output=True, text=True)

print("код завершення:", short_run.returncode)
print("останній рядок помилки:", short_run.stderr.strip().splitlines()[-1])
print()
print("а це те, що людина встигла побачити:", repr(short_run.stdout))

assert "EOFError" in short_run.stderr
print()
print("✅ скільки викликів input(), стільки рядків має бути на вході — інакше EOFError")
````

- `int()` відкидає дробову частину, а `round()` округлює — це різні дії;
- `int("сорок")` дає `ValueError`, і полагодити це можна буде лише в темі 19;
- порівняння без `.strip().lower()` дає `False` на однакових на вигляд рядках;
- наш розбір числа на розряди зрізами дає те саме, що `f"{n:,}"`;
- наше вирівнювання пробілами дає те саме, що `format(item, "<16")`;
- **справжній `input()` працює** — коли скрипт запускають окремим процесом і подають
  йому відповіді наперед.

Головне: усередині міні-мови формату немає магії. Там ті самі кілька дій із рядками,
